```python
#Train and Save Your Model
#Here we are training a machine learning model using scikit-learn and then save 
#it with BentoML’s save_model function to create a reusable model artifact, 
#create train.py and paste the below code.
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
import bentoml
# Fake training dataset
data = pd.DataFrame({
    "monthly_charges": [50, 80, 20, 90, 60, 100, 30, 70],
    "tenure": [12, 2, 24, 1, 10, 3, 30, 5],
    "support_tickets": [1, 5, 0, 7, 2, 6, 0, 4],
    "churn": [0, 1, 0, 1, 0, 1, 0, 1]
})
X = data[["monthly_charges", "tenure", "support_tickets"]]
y = data["churn"]
model = LogisticRegression()
model.fit(X, y)
# Save model into BentoML model store
bentoml.sklearn.save_model("churn_model", model)
print("Model trained and saved successfully!")



```

```python
import bentoml
import numpy as np

@bentoml.service(resources={"cpu": 1})
class ChurnPredictor:

    def __init__(self):
        self.model = bentoml.sklearn.load_model("churn_model:latest")

    @bentoml.api()
    def predict(self, monthly_charges: float, tenure: int, support_tickets: int) -> dict:
        features = np.array([[monthly_charges, tenure, support_tickets]])
        prediction = self.model.predict(features)[0]
        probability = self.model.predict_proba(features)[0][1]

        return {
            "churn_prediction": int(prediction),
            "churn_probability": float(probability)
        }
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import bentoml

# ────────────────────────────────────────────────
# 1. Load data (tab-separated marketing_campaign.csv)
# ────────────────────────────────────────────────
print("Loading data...")
df = pd.read_csv(
    r"C:\Users\KIIT\Downloads\archive (1)\marketing_campaign.csv",
    sep="\t",
    encoding="utf-8"
)

# Keep only numeric columns
df_numeric = df.select_dtypes(include=[np.number])

print(f"Numeric columns found: {list(df_numeric.columns)}")
print(f"Shape before cleaning: {df_numeric.shape}")

# ────────────────────────────────────────────────
# 2. Create and fit preprocessing pipeline steps
# ────────────────────────────────────────────────
print("Fitting imputer...")
imputer = SimpleImputer(strategy='mean')
imputer.fit(df_numeric)

print("Fitting scaler...")
scaler = StandardScaler()
df_imputed = imputer.transform(df_numeric)
scaler.fit(df_imputed)

print("Fitting PCA...")
pca = PCA(n_components=3)
df_scaled = scaler.transform(df_imputed)
pca.fit(df_scaled)

# Optional: see how much variance is explained
print(f"Explained variance ratio: {pca.explained_variance_ratio_}")
print(f"Total explained variance: {sum(pca.explained_variance_ratio_):.4f}")

# ────────────────────────────────────────────────
# 3. Save each model separately (modern BentoML style)
# ────────────────────────────────────────────────
print("\nSaving models to BentoML model store...")

bentoml.sklearn.save_model(
    name="marketing_imputer",
    model=imputer,
    metadata={
        "strategy": "mean",
        "description": "Imputes missing values with column mean - marketing dataset"
    }
)

bentoml.sklearn.save_model(
    name="marketing_scaler",
    model=scaler,
    metadata={
        "description": "StandardScaler fitted on imputed marketing campaign numeric data"
    }
)

bentoml.sklearn.save_model(
    name="marketing_pca",
    model=pca,
    metadata={
        "n_components": 3,
        "description": "PCA reducing to 3 components after scaling"
    }
)

print("\nAll models saved successfully!")
print("You can now list them with:  bentoml models list")
print("Example tags: marketing_imputer:latest, marketing_scaler:latest, marketing_pca:latest")
# Quick verification
loaded_imputer = bentoml.sklearn.get("marketing_imputer:latest").load_model()
loaded_scaler  = bentoml.sklearn.get("marketing_scaler:latest").load_model()
loaded_pca     = bentoml.sklearn.get("marketing_pca:latest").load_model()
print("\nVerification:")
print("Imputer loaded successfully →", isinstance(loaded_imputer, SimpleImputer))
print("Scaler loaded successfully  →", isinstance(loaded_scaler, StandardScaler))
print("PCA loaded successfully     →", isinstance(loaded_pca, PCA))
# service.py
import bentoml
from bentoml.io import JSON
import numpy as np

imputer_runner = bentoml.sklearn.get("marketing_imputer:latest").to_runner()
scaler_runner  = bentoml.sklearn.get("marketing_scaler:latest").to_runner()
pca_runner     = bentoml.sklearn.get("marketing_pca:latest").to_runner()

svc = bentoml.Service("marketing-pca-pipeline")

input_spec = JSON(pydantic_model=dict)  # or create proper pydantic model

@svc.api(input=input_spec, output=JSON())
async def predict(payload: dict) -> dict:
    # Expecting: {"values": [1.2, 3.4, null, 567, ...]}  ← same order as training columns
    try:
        arr = np.array(payload["values"], dtype=object).reshape(1, -1)
    except Exception as e:
        return {"error": str(e)}

    imputed = await imputer_runner.predict.async_run(arr)
    scaled  = await scaler_runner.transform.async_run(imputed)
    reduced = await pca_runner.transform.async_run(scaled)

    return {
        "pca_components": reduced[0].tolist(),
        "explained_variance": [round(x, 4) for x in pca_runner.model.explained_variance_ratio_]
    }
```

```python
#bentoml models list
# service.py
import bentoml
from bentoml.io import JSON
import numpy as np
from pydantic import BaseModel, Field
from typing import List, Any
# ────────────────────────────────────────────────
# Input / Output schemas (validation + auto OpenAPI docs)
# ────────────────────────────────────────────────
class PredictInput(BaseModel):
    values: List[Any] = Field(
        ...,
        description="List of values in the SAME ORDER as the numeric columns used during training",
        examples=[[45, 1200, 3, None, 890, 56.2, 1, 2.3, 4.5]]
    )
class PredictOutput(BaseModel):
    pca_components: List[float]
    explained_variance_ratios: List[float]
    message: str = "PCA transformation completed"
# ────────────────────────────────────────────────
# Load models once at startup (recommended modern style)
# ────────────────────────────────────────────────
try:
    imputer = bentoml.sklearn.get("marketing_imputer:latest").load_model()
    scaler  = bentoml.sklearn.get("marketing_scaler:latest").load_model()
    pca     = bentoml.sklearn.get("marketing_pca:latest").load_model()
    print("All models loaded successfully")
except Exception as e:
    print(f"Error loading models: {e}")
    raise
# ────────────────────────────────────────────────
# BentoML Service
# ────────────────────────────────────────────────
svc = bentoml.Service(
    name="marketing-pca-pipeline",
    # description="Transforms marketing campaign data → 3 PCA components"
)
input_spec  = JSON(pydantic_model=PredictInput)
output_spec = JSON(pydantic_model=PredictOutput)
@svc.api(input=input_spec, output=output_spec)
async def predict(input_data: PredictInput) -> PredictOutput:
    """
    Transform input features → impute missing → scale → PCA (3 components)
    """
    try:
        # Convert to 2D numpy array (1 sample, n_features)
        X = np.array(input_data.values, dtype=np.float64).reshape(1, -1)
    except Exception as e:
        raise ValueError(
            f"Input conversion failed: {str(e)}\n"
            f"Expected a list of numbers (or numbers + null). Got: {input_data.values}"
        )
    # Pipeline steps
    X_imputed = imputer.transform(X)
    X_scaled  = scaler.transform(X_imputed)
    X_pca     = pca.transform(X_scaled)
    return PredictOutput(
        pca_components = X_pca[0].tolist(),
        explained_variance_ratios = [round(v, 4) for v in pca.explained_variance_ratio_],
        message = "Success"
    )
```
